# 06: Summary: Correlation Matrix & Key Numbers

Closing notebook: one correlation heatmap that answers "what matters most" at a
glance, plus a printed summary of the headline numbers to paste into the final
report.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 110

df = pd.read_csv("../data/processed/delivered_orders.csv")
```

## Correlation heatmap

Recompute `freight_ratio` fresh here (rather than reusing the filtered frame
from notebook 03) so this matrix is self-consistent and reproducible on its
own.

```python
df["freight_ratio"] = df["total_freight"] / df["total_price"]
df = df[df["freight_ratio"] < 5]

corr_cols = ["review_score", "delivery_delay_days", "delivery_time_days", "freight_ratio", "total_price"]
corr_matrix = df[corr_cols].dropna().corr()

fig, ax = plt.subplots(figsize=(7, 6))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0, ax=ax)
ax.set_title("Correlation Matrix: What Drives Review Score?")
plt.savefig("../reports/figures/07_correlation_heatmap.png", bbox_inches="tight")
plt.show()
```

## Review score distribution (baseline chart)

Establishes the U-shape referenced throughout the report; worth having in this
summary notebook even though it was first shown in notebook 03's underlying
data, since this is the notebook someone skims last.

```python
fig, ax = plt.subplots()
order = [1, 2, 3, 4, 5]
counts = df["review_score"].value_counts().reindex(order)
bars = ax.bar(counts.index.astype(str), counts.values, color=sns.color_palette("RdYlGn", 5))
ax.set_xlabel("Review Score")
ax.set_ylabel("Number of Orders")
ax.set_title("Distribution of Review Scores")
for bar, val in zip(bars, counts.values):
    ax.text(bar.get_x() + bar.get_width()/2, val + 500, f"{val:,.0f}", ha="center", fontsize=9)
plt.savefig("../reports/figures/01_review_distribution.png", bbox_inches="tight")
plt.show()
```

## Headline numbers for the report



```python
print(f"Total delivered orders analyzed: {len(df):,}")
print(f"Average review score: {df['review_score'].mean():.2f}")
print(f"% orders delivered late: {(df['is_late'].mean() * 100):.1f}%")
print(f"Avg review score, on-time: {df[df['is_late']==False]['review_score'].mean():.2f}")
print(f"Avg review score, late:    {df[df['is_late']==True]['review_score'].mean():.2f}")
```
